```python
import pandas as pd

# Carregue seu arquivo
df = pd.read_pickle("data_strava_activities.pkl")
```

```python
df['type'] = df['type'].apply(lambda x: x.root)
```

```python
df = df.loc[df['type'] == 'Run'].copy()
```

```python
df['date'] = pd.to_datetime(df['date'])
```

```python
# Extract components
df['year']  = df['date'].dt.year
df['month'] = df['date'].dt.month
df['day']   = df['date'].dt.day
df['hour']  = df['date'].dt.hour
df['min']   = df['date'].dt.minute
```

```python
df['weekday_num'] = df['date'].dt.weekday
```

```python
df['weekday_name'] = df['date'].dt.day_name()
```

```python
df['month_name'] = df['date'].dt.month_name()
```

```python
import numpy as np

def standardize_all(values):
    return np.round(values).astype(int)
```

```python
df['distance_km'] = df['distance_km'].apply(lambda x: standardize_all(x))
```

```python
df = df.loc[df['pace_min_km']>2]
```

```python
# Converta para JSON (orientado a registros é o mais fácil para JS)
# Salve dentro da pasta 'static/data/' do seu projeto Hugo
df.to_json("running_data.json", orient="records", date_format="iso")
```
